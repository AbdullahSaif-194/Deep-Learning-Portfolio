```python
data['voulume'] = data['x']*data['y']*data['z']
dropcolumns = ['x','y','z']
data.drop(dropcolumns,axis = 1, inplace = True)
data.head()
```

```python
datas = sns.load_dataset('diamonds')
print(f"data: {len(data)}   &&  datas:  {len(datas)}")
data['carat'] = datas['carat']
data['cut'] = datas['cut']
data['color'] = datas['color']

print(f"\n/nta: {len(data)}   &&  datas:  {len(datas)}")
```

```python
print((data['voulume'] == 0).sum())
```

```python
# we will remove the zero columns as they are just 19 0.04% of the whole data so removing them is better
# Keep only the rows where volume is greater than 0
data = data[data['voulume'] > 0]

# Verify they are gone
print(f"Remaining rows: {len(data)}")
print(f"Zero volume rows left: {(data['voulume'] == 0).sum()}")
```

```python
# 1. Calculate IQR for Volume
Q1 = data['voulume'].quantile(0.25)
Q3 = data['voulume'].quantile(0.75)
IQR = Q3 - Q1

# 2. Define Bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

```python
# Calculate outliers
outliers = data[(data['voulume'] < lower_bound) | (data['voulume'] > upper_bound)]

print(f"Total rows before: {len(data)}")
print(f"Number of rows to be dropped: {len(outliers)}")
print(f"Percentage of data lost: {round(len(outliers)/len(data)*100, 2)}%")
```

```python
# Only keep rows inside the bounds
data = data[(data['voulume'] >= lower_bound) & (data['voulume'] <= upper_bound)]
```

```python
Q1_t = data['table'].quantile(0.25)
Q3_t = data['table'].quantile(0.75)
IQR_t = Q3_t-Q1_t
lower_bound_t = Q1_t-1.5*IQR_t
upper_bound_t = Q3_t + 1.5*IQR_t
outliers = data[(data['table']<lower_bound_t) | (data['table'] > upper_bound_t)]
print(f"Total rows before : {len(data)}")
print(f"Number of rows to be dropped: {len(outliers)}")
print(f"Percentage of data lost: {round(len(outliers)/len(data)*100,2)}%")
```

```python
data = data[(data['table']>=lower_bound_t) & (data['table'] <= upper_bound_t)]
print(f"{len(data)}")
```

```python
# 1. Calculate IQR for Depth
Q1_d = data['depth'].quantile(0.25)
Q3_d = data['depth'].quantile(0.75)
IQR_d = Q3_d - Q1_d

lower_d = Q1_d - 1.5 * IQR_d
upper_d = Q3_d + 1.5 * IQR_d

# 2. Filter using AND (&) to keep the middle range
data = data[(data['depth'] >= lower_d) & (data['depth'] <= upper_d)]

print(f"Final dataset size after all cleaning: {len(data)}")
```

```python
from sklearn.preprocessing import LabelEncoder
# Initialize a dictionary to store enco ders (useful for inverse transforms later)
label_encoders = {}

for col in ['cut', 'color','clarity']:
    le = LabelEncoder()
    data[col] = le.fit_transform(data[col].astype(str))  # Convert to string to handle NaN
    label_encoders[col] = le  # Store the encoder for future use

```

```python
# Define the order of quality
cut_map = {"Fair": 1, "Good": 2, "Very Good": 3, "Premium": 4, "Ideal": 5}
clarity_map = {"I1": 1, "SI2": 2, "SI1": 3, "VS2": 4, "VS1": 5, "VVS2": 6, "VVS1": 7, "IF": 8}
color_map = {"J": 1, "I": 2, "H": 3, "G": 4, "F": 5, "E": 6, "D": 7}

# Apply the mapping
data['cut'] = data['cut'].map(cut_map)
data['clarity'] = data['clarity'].map(clarity_map)
data['color'] = data['color'].map(color_map)
```

```python
from sklearn.model_selection import train_test_split

# Define Features and Target
X = data.drop('price', axis=1)
y = data['price']

# Split: 80% for training, 20% for testing
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=101)

print(f"Training set size: {X_train.shape}")
print(f"Testing set size: {X_test.shape}")
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression

# Initialize the scaler
scaler = StandardScaler()

# 1. Fit AND Transform the training data
X_train_scaled = scaler.fit_transform(X_train)

# 2. ONLY Transform the test data (using the mean/std from the training data)
X_test_scaled = scaler.transform(X_test)


model = LinearRegression()
# Now train the model using the scaled data
model.fit(X_train_scaled, y_train)
y_pred = model.predict(X_test_scaled)
```

```python
# Calculate the performance
from sklearn.metrics import mean_squared_error, r2_score

r2 = r2_score(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse) # This turns 1,294,264 into ~1,137

print(f"R2 Score: {r2:.4f}")
print(f"Actual Average Error (RMSE): ${rmse:.2f}")
```

```python
# Create a dataframe to see the importance of each feature
coefficients = pd.DataFrame({'Feature': X.columns, 'Importance': model.coef_})
coefficients = coefficients.sort_values(by='Importance', ascending=False)

print(coefficients)
```

```python
# Combine Actual and Predicted values into a table
comparison_df = pd.DataFrame({
    'Actual Price': y_test.values,
    'Predicted Price': y_pred.round(2),
    'Difference (Error)': (y_test.values - y_pred).round(2)
})

print(comparison_df.head(100))
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x=y_test, y=y_pred, alpha=0.5)

# Add the "Perfect Prediction" line
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], color='red', lw=2, linestyle='--')

plt.xlabel('Actual Price ($)')
plt.ylabel('Predicted Price ($)')
plt.title('Actual vs. Predicted Diamond Prices')
plt.show()
```

```python
# Plotting the errors
sns.histplot(comparison_df['Difference (Error)'], kde=True)
plt.title('Distribution of Prediction Errors')
plt.show()
```
